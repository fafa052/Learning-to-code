```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load image \
image_path = '/content/cohete.jpg' #image path
image = cv2.imread(image_path)

# Convert image to RGB to convert the image must use "cv2.COLOR_BGR2RGB" \
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Define kernels for different filters \
kernels = {
    'Identity': np.array([[0,0,0],[0,1,0],[0,0,0]]), # Identity kernel \
    'Edge Detection': np.array([[-1,-1,-1],[-1,8,-1],[-1,-1,-1]]), # Edge detection Kernel \
    'Gaussian Blur': np.array([[1,2,1],[2,4,2],[1,2,1]])/16, # Gaussian Blur kernel \
    'Emboss': np.array([[-2,-1,0],[-1,1,1],[0,1,2]]) # Emboss kernel \
}

# Apply each kernel using convolution \
filtered_images = {}
for filter_name,kernel in kernels.items():
  filtered_image = cv2.filter2D(image_rgb, -1, kernel)
  filtered_images[filter_name] = filtered_image

# Plot the original and filtered images \
plt.figure(figsize=(10,10))
plt.subplot(3,2,1)
plt.imshow(image_rgb)
plt.title('Original Image')

for i, (filter_name, filtered_image) in enumerate(filtered_images.items(), start =2):
  plt.subplot(3,2,i)
  plt.imshow(filtered_image)
  plt.title(filter_name)
  plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Step 1: Load the image
# We read the image in grayscale mode for simpler edge detection
image = cv2.imread('/content/Peresozo.jpg', cv2.IMREAD_GRAYSCALE)

# Step 2: Apply Gaussian Blur
# GaussianBlur is used to reduce noise and detail in the image before performing edge detection.
# The kernel size (5,5) indicates a 5x5 window, and the '0' tells OpenCV to calculate the standard deviation automatically
blurred = cv2.GaussianBlur(image, (5,5),0)

# Step 3: Apply Canny Edge Detection
# The Canny algorithm uses the blurred image as input and the lower (50) and upper (150) thresholds for detecting edges
# Pixels with a gradient between the thresholds will be  included if connected to strong edges.
edges = cv2.Canny(blurred,100,150)

color_image = cv2.imread('/content/Peresozo.jpg')
#color_edges = cv2.cvtColor(edges, cv2.COLOR_GRAY2BGR)
image_rgb = cv2.cvtColor(color_image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10,5))

# Original Image
plt.subplot(2,2,1)
plt.imshow(image_rgb)
plt.title('Original Image')
plt.axis("off")

# Grayscale Image
plt.subplot(2,2,2)
plt.imshow(image,cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")

# Blurred Image
plt.subplot(2,2,3)
plt.imshow(blurred, cmap='gray')
plt.title("Blurred Image")
plt.axis("off")

# Edge Detection Image
plt.subplot(2,2,4)
plt.imshow(edges, cmap='gray')
plt.title("Canny Edge Detection Image")
plt.axis("off")

plt.tight_layout()
plt.show()


```
